```python
SBL = StatsBombLoader()
df_events = SBL.events(game_id=3788741)

df_actions = (
    spadl.add_names(df_actions)  # add actiontype and result names
    .merge(SBL.teams(game_id=3788741))  # add team names
    .merge(SBL.players(game_id=3788741))  # add player names
)
df_actions.head()
```

```python
import socceraction.atomic.spadl as atomicspadl

df_atomic = atomicspadl.convert_to_atomic(df_actions)
df_atomic.head(10)
```

```python
import pandas as pd

df_competitions = api.competitions()

all_games = []
top_games = []
for row in df_competitions.itertuples(index=False):
   # print(row)
    df_games = api.games(competition_id=row.competition_id, season_id=row.season_id)
    all_games.append(df_games)

    if row.competition_name in top_leagues:
        df_top_games = api.games(
            competition_id=row.competition_id, season_id=row.season_id
        )
        top_games.append(df_top_games)

# Combine all results into a single DataFrame
df_all_games = pd.concat(all_games, ignore_index=True)
df_top_games = pd.concat(top_games, ignore_index=True)
```
