# 03 — Train / val / test split

Split `masked_df.csv` by `game_id` (no game spans two sets), shuffled with
`random_state=42`, 80 / 10 / 10. The `-2` boundary marker in `target` is
remapped to `106` first. Writes `train_set.csv` / `val_set.csv` / `test_set.csv`.

```python
import pandas as pd
```

```python
df = pd.read_csv("data/masked_df.csv", index_col=0)
df.loc[df["target"] == -2, "target"] = 106

ml_df = df
```

```python
ml_events = ml_df[["game_id"]].drop_duplicates()
events_shuffled = ml_events.sample(frac=1, random_state=42)
n1 = int(0.8 * len(events_shuffled))
n2 = int(0.9 * len(events_shuffled))
```

```python
df_shuffled1 = pd.merge(events_shuffled.iloc[:n1], ml_df, on=["game_id"], how="left")
df_shuffled2 = pd.merge(events_shuffled.iloc[n1:n2], ml_df, on=["game_id"], how="left")
df_shuffled3 = pd.merge(events_shuffled.iloc[n2:], ml_df, on=["game_id"], how="left")
```

```python
df_shuffled1.to_csv("data/train_set.csv", index=False)
df_shuffled2.to_csv("data/val_set.csv", index=False)
df_shuffled3.to_csv("data/test_set.csv", index=False)

print(df_shuffled1.shape, df_shuffled2.shape, df_shuffled3.shape)
```
