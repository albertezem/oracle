```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
%matplotlib inline
import random

random.seed(42)
```

```python
df_shuffled = ml_df.sample(frac=1)
n1 = int(0.8 * len(ml_df))
n2 = int(0.9 * len(ml_df))
```

```python
def build_dataset(df):
    X = df.drop(['target'], axis=1)
    X = torch.tensor(X.values).float()

    Y = df['target']
    Y = torch.tensor(Y.values).float()
    Y = F.one_hot(Y.long(), num_classes=106)

    return X, Y
```

```python
Xtr, Ytr = build_dataset(df_shuffled.iloc[:n1])
Xval, Yval = build_dataset(df_shuffled.iloc[n1:n2])
Xte, Yte = build_dataset(df_shuffled.iloc[n2:])
```

```python
Xtr.shape
```

```python
n_hidden = 80
vocab_size = len(Ytr.unique())

W1 = torch.randn((Xtr.shape[1], n_hidden))
b1 = torch.randn(n_hidden)
W2 = torch.randn((n_hidden, vocab_size))
b2 = torch.randn(vocab_size)

parameters = [W1, b1, W2, b2]

for p in parameters:
    p.requires_grad = True
```

```python
max_steps = 10000
batch_size = 32
lr = 1
lossi = []

for i in range(max_steps):
    ix = torch.randint(0, W1.shape[0], (batch_size,))
    Xb, Yb = Xtr[ix], Ytr[ix]
    
    #forward pass
    
    hpreact = Xb @ W1 + b1
    h = torch.relu(hpreact)
    
    logits = h @ W2 + b2
    
    loss = F.binary_cross_entropy_with_logits(logits, Ytr[ix])
    
    # backward pass
    
    for p in parameters:
        p.grad = None
        
    loss.backward()
    
    for p in parameters:
        p.data += -lr * p.grad
        
    print(loss)
    
    
```
