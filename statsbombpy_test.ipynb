```python
events['is_home'] = (events['home_team'] == events['team'])
events["is_yellow"] = (events["foul_committed_card"] == "Yellow Card") | (events.get("bad_behaviour_card") == "Yellow Card")
events['home_yellows'] = (events['is_home'] & events['is_yellow'])
events["away_yellows"] = ~events["is_home"] & events["is_yellow"]
events['home_yc_cumul'] = events['home_yellows'].cumsum().shift(1, fill_value=0)
events["away_yc_cumul"] = events["away_yellows"].cumsum().shift(1, fill_value=0)
events[events["is_yellow"]]
```

```python
events["is_red"] = (
    (events["foul_committed_card"] == "Red Card")
    | (events["foul_committed_card"] == "Second Yellow")
    | (events.get("bad_behaviour_card") == "Red Card")
    | (events.get("bad_behaviour_card") == "Second Yellow")
)
events["home_reds"] = events["is_home"] & events["is_red"]
events["away_reds"] = ~events["is_home"] & events["is_red"]
events["home_rc_cumul"] = events["home_reds"].cumsum().shift(1, fill_value=0)
events["away_rc_cumul"] = events["away_reds"].cumsum().shift(1, fill_value=0)

events = events.copy()
```

```python
events['shot_outcome'].unique()
```

```python
events["is_goal"] = (events["shot_outcome"] == "Goal") | (events["type"] == "Own Goal For")
events["home_goals"] = events["is_home"] & events["is_goal"]
events["away_goals"] = ~events["is_home"] & events["is_goal"]
events["hgoals_cumul"] = events["home_goals"].cumsum().shift(1, fill_value=0)
events["agoals_cumul"] = events["away_goals"].cumsum().shift(1, fill_value=0)
events.tail()
```
